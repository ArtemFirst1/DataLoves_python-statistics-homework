**Завдання 3.** Для того ж набору даних в змінній `sum_purchases` обчисліть **медіану** кожним з зазначених методів:  

  3.1. без використання жодних бібліотек, користуючись тільки Python.  
  3.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
# 3.1. без використання жодних бібліотек, користуючись тільки Python
otsortirovanye_pokepochki = sorted(sum_purchases)
n = len(otsortirovanye_pokepochki)

if n % 2 == 0:
    median_pythonchik = (otsortirovanye_pokepochki[n // 2 - 1] + otsortirovanye_pokepochki[n // 2]) / 2
else:
    median_pythonchik = otsortirovanye_pokepochki[n // 2]

print("Медіана (Python):", median_pythonchik)

# 3.2.  використовуючи відповідний метод бібліотеки numpy
median_numpy = np.median(sum_purchases)
print("Медіана (NumPy):", median_numpy)

# Порівняння з вибірковим середнім
mean_purchases = np.mean(sum_purchases)

if mean_purchases > median_numpy:
    print("Середнє більше за медіану")
elif mean_purchases < median_numpy:
    print("Медіана більша за середнє")
else:
    print("Середнє та медіана рівні")

print("Різниця між середнім та медіаною:", abs(mean_purchases - median_numpy))
print("Вибіркове середнє:", mean_purchases)
print("Медіана:", median_numpy)
print("Результати медіани співпадають:", np.isclose(median_pythonchik, median_numpy))
```

**Вибіркове середнє я б використовував для визначення середніх витрат клієнтів та аналізу загального обсягу покупок.**

**Медіану я б використовував для визначення типових витрат клієнтів, оскільки вона менш чутлива до дуже великих або маленьких покупок.**

**Якщо середнє значно більше за медіану, це може означати, що деякі клієнти витрачають набагато більше за інших. Для аналізу типових витрат клієнта медіана часто буде більш показовою**

**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:  

  4.1. без використання жодних бібліотек, користуючись тільки Python.  
  4.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран.

```python
# 4.1.  без використання жодних бібліотек, користуючись тільки Python.
n = len(sum_purchases)
mean_pythonchik = sum(sum_purchases) / n

variance_pythonchik = sum((x - mean_python)**2 for x in sum_purchases) / (n - 1)
std_python = variance_pythonchik**0.5

print("Стандартне відхилення (Python):", std_python)

# 4.2.  використовуючи відповідний метод бібліотеки numpy.
std_numpy = np.std(sum_purchases, ddof=1)

print("Стандартне відхилення (NumPy):", std_numpy)

# Порівняння результатів
print("Результати співпадають:", np.isclose(std_python, std_numpy))
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
# 6.1. без використання жодних бібліотек, користуючись тільки Python.

mean_x = sum(sum_purchases) / len(sum_purchases)
mean_y = sum(visits) / len(visits)

numerator = sum((x - mean_x) * (y - mean_y) for x, y in zip(sum_purchases, visits))

denominator_x = sum((x - mean_x)**2 for x in sum_purchases)
denominator_y = sum((y - mean_y)**2 for y in visits)

correlation_python = numerator / (denominator_x * denominator_y)**0.5

print("Кореляція Пірсона (Python):", correlation_python)
```

```python
# 6.2. використовуючи відповідний метод бібліотеки numpy.
correlation_numpy = np.corrcoef(sum_purchases, visits)[0, 1]

print("Кореляція Пірсона (NumPy):", correlation_numpy)

# Порівняння результатів
print("Результати співпадають:", np.isclose(correlation_python, correlation_numpy))
```

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
# Обчислення порогу викидів
threshold = np.mean(sum_purchases) + 2 * np.std(sum_purchases)

print("Поріг викидів:", threshold)

# Створення маски True/False (скористався підказкою в чаті, хотів робити через for)
mask = np.array(sum_purchases) <= threshold

# Видалення викидів з обох масивів
clean_purchases = np.array(sum_purchases)[mask]
clean_visits = np.array(visits)[mask]

print("Кількість записів до очищення:", len(sum_purchases))
print("Кількість записів після очищення:", len(clean_purchases))

# Обчислення кореляції Пірсона
correlation_clean = np.corrcoef(clean_purchases, clean_visits)[0, 1]

print("Кореляція без викидів:", correlation_clean)

# Побудова графіка
plt.scatter(clean_purchases, clean_visits)

plt.title("Scatter Plot Without Outliers")
plt.xlabel("Sum of Purchases")
plt.ylabel("Number of Visits");
```

**Після видалення 2 викидів кількість спостережень зменшилася з 20 до 18. Коефіцієнт кореляції Пірсона становить −0,3685, що свідчить про помірний негативний лінійний зв'язок між кількістю відвідувань та сумою покупок.**

**На графіку видно, що зі збільшенням кількості відвідувань сума покупок має тенденцію до зменшення. Однак зв'язок не є сильним, тому однозначно стверджувати про залежність між цими показниками не можна.**
